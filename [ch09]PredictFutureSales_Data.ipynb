## 데이터 병합

```python
train = sales_train.merge(shops, on='shop_id', how='left')
train = train.merge(items, on='item_id', how='left')
train = train.merge(item_categories, on='item_category_id', how='left')

train.head()
```

# 피처 요약 표 만들기

```python
def resumetable(df):
  print(f'데이터셋 형상: {df.shape}')
  summary = pd.DataFrame(df.dtypes, columns=['데이터 타입'])
  summary = summary.reset_index()
  summary = summary.rename(columns={'index': '피처'})
  summary['결측값 개수'] = df.isnull().sum().values
  summary['고윳값 개수'] = df.nunique().values
  summary['첫 번째 값'] = df.loc[0].values
  summary['두 번째 값'] = df.loc[1].values

  return summary
```

```python
resumetable(train)

# 각 id와 name은 1:1 매칭 (shop_id, shop_name), (item_id, item_name), (item_category_id, item_category_name)
# => id와 name 중 하나만 있어도 됨 (둘 중 하나 제거)
# => name 피처를 활용해 파생 피처를 만들어 성능 개선 가능
```

# 데이터 시각화

```python
import seaborn as sns
import matplotlib as mpl
import matplotlib.pyplot as plt
%matplotlib inline
```

## 일별 판매량

```python
sns.boxplot(y='item_cnt_day', data=train)
```

## 그룹화

```python
group = train.groupby('date_block_num').agg({'item_cnt_day': 'sum'})
group.reset_index() # 인덱스 재설정
```

## 월별 판매량

```python
mpl.rc('font', size=13)
figure, ax = plt.subplots()
figure.set_size_inches(11, 5)

# 월별 총 상품 판매량
group_month_sum = train.groupby('date_block_num').agg({'item_cnt_day': 'sum'})
group_month_sum = group_month_sum.reset_index()

# 월별 총 상품 판매량 막대 그래프
sns.barplot(x='date_block_num', y='item_cnt_day', data=group_month_sum)
ax.set(title='Distribution of monthly item counts by date block numbere',
       xlabel='Date block number',
       ylabel='Monthly item counts')
```

## 상품분류별 판매량

```python
train['item_category_id'].nunique()
```

```python
figure, ax = plt.subplots()
figure.set_size_inches(11, 5)

# 상품분류별 총 상품 판매량
group_cat_sum = train.groupby('item_category_id').agg({'item_cnt_day': 'sum'})
group_cat_sum = group_cat_sum.reset_index()

# 월간 판매량이 10,000개를 초과하는 상품분류만 추출
group_cat_sum = group_cat_sum[group_cat_sum['item_cnt_day'] > 10000]

# 상품분류별 총 상품 판매량 막대 그래프
sns.barplot(x='item_category_id', y='item_cnt_day', data=group_cat_sum)
ax.set(title='Distribution of total item counts by item category id',
       xlabel='Date block number',
       ylabel='Total item counts')
ax.tick_params(axis='x', labelrotation=90) # x축 라벨 회전
```

## 상점별 판매량

```python
figure, ax = plt.subplots()
figure.set_size_inches(11, 5)

group_cat_sum = train.groupby('shop_id').agg({'item_cnt_day': 'sum'})
group_cat_sum = group_cat_sum.reset_index()

group_cat_sum = group_cat_sum[group_cat_sum['item_cnt_day'] > 10000]

sns.barplot(x='shop_id', y='item_cnt_day', data=group_cat_sum)
ax.set(title='Distribution of total item counts by item shop id',
       xlabel='Date block number',
       ylabel='Total item counts')
ax.tick_params(axis='x', labelrotation=90)
```
